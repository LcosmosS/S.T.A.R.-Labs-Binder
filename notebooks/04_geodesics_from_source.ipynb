Great circles are geodesics on a unit sphere with its standard induced metric. This is an exact classical geometry example, **not** an RTCH entropy metric or spacetime solution.

```python
import numpy as np
import matplotlib.pyplot as plt
t = np.linspace(0, 2*np.pi, 400)
u = np.linspace(0, 2*np.pi, 40)
v = np.linspace(0, np.pi, 24)
U, V = np.meshgrid(u, v)
fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(projection="3d")
ax.plot_surface(np.cos(U)*np.sin(V), np.sin(U)*np.sin(V), np.cos(V), alpha=0.15)
for phase in [0.0, 0.4, 0.8]:
    p = np.array([np.cos(t), np.sin(t)*np.cos(phase), np.sin(t)*np.sin(phase)])
    assert np.allclose(np.sum(p*p, axis=0), 1.0)
    ax.plot(*p, label=f"great circle {phase:.1f}")
ax.set_box_aspect((1, 1, 1))
ax.legend()
plt.show()
print("Unit-sphere geodesic example completed; not a spacetime calculation.")
```
