We define an explicit Gaussian scalar field. The word *entropy* is only a label for this toy function; it is not thermodynamic entropy. Its gradient and Hessian follow analytically.

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-2, 2, 101)
X, Y = np.meshgrid(x, x)
M = np.exp(-(X**2 + Y**2)/2)
dMx, dMy = -X*M, -Y*M
laplacian = (X**2 + Y**2 - 2)*M
assert np.isfinite(laplacian).all()
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].contourf(X, Y, M, levels=16)
ax[0].set_title("Synthetic scalar field M")
ax[1].contourf(X, Y, laplacian, levels=16)
ax[1].set_title("Analytic Laplacian of M")
plt.tight_layout()
plt.show()
print("Analytic derivatives computed; no thermodynamic interpretation.")
```
