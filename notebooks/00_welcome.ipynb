# S.T.A.R.-Labs Binder
This is an illustrative numerical demonstration, **not** a controlled experiment or evidence for a scientific hypothesis.

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 2*np.pi, 200)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, np.sin(x))
ax.set(xlabel='x', ylabel='sin(x)', title='Binder smoke test')
plt.show()
assert np.isfinite(np.sin(x)).all()
print('Binder smoke test passed')
```
