A sampled circle is perturbed with a fixed random seed. We compare pairwise-distance changes, **not** persistent homology, Betti numbers, or Wasserstein distances. The full source TDA pipeline is deliberately not reproduced.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(99)
t = np.linspace(0, 2*np.pi, 32, endpoint=False)
base = np.column_stack([np.cos(t), np.sin(t)])
noise_levels = np.linspace(0, 0.12, 13)
D0 = np.linalg.norm(base[:, None, :] - base[None, :, :], axis=-1)
errors = []
for sigma in noise_levels:
    perturbed = base + rng.normal(0, sigma, size=base.shape)
    D = np.linalg.norm(perturbed[:, None, :] - perturbed[None, :, :], axis=-1)
    errors.append(np.max(np.abs(D-D0)))
assert errors[0] == 0 and np.all(np.isfinite(errors))
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(noise_levels, errors, marker="o")
ax.set(xlabel="synthetic noise standard deviation", ylabel="maximum distance-matrix change", title="Illustrative metric sensitivity (not persistent homology)")
plt.show()
print("Distance stability illustration completed; no topological inference.")
```
